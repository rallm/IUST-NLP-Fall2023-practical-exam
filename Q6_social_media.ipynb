#Social Media Workshop


Determine the sentiments of the first 200 samples from the Reuters corpus and create a plot. Employ the methodology you utilized in the social media workshop.

```python
# Write your code here.
from transformers import pipeline
import matplotlib.pyplot as plt
from nltk.corpus import reuters
from collections import Counter

# Load sentiment analysis pipeline
classifier = pipeline("sentiment-analysis")

# Get the first 200 samples from Reuters
raw_data = [reuters.raw(fileid) for fileid in reuters.fileids()[:200]]

# Predict sentiments (truncating to fit model limits)
results = classifier(raw_data, truncation=True)
labels = [res['label'] for res in results]

# Plotting the results
counts = Counter(labels)
plt.bar(counts.keys(), counts.values())
plt.title("Sentiment Analysis of first 200 Reuters samples")
plt.xlabel("Sentiment")
plt.ylabel("Count")
plt.show()
```
